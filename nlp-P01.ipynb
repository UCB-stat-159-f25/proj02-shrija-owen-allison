### Read Data

```python
# imports
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-dark') 
```

```python
sou = pd.read_csv("data/SOTU.csv")
sou
```

### Exploratory Data Analysis

Replicate the plots below using the hints specified. For each plot, provide some commentary describing the results/anything interesting you might see.

#### Number of Speeches per President

```python
# Counts
counts = sou["President"].value_counts()
# Fix Order
counts = counts.reindex(sou["President"].drop_duplicates())

# Plot
plt.figure(figsize=(12,6))
counts.plot(kind="bar")
plt.title("Number of Speeches per President")
plt.xlabel("President")
plt.ylabel("Count")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show();
```

#### Commentary

It was interesting to me that for the most part, each President had a speech per year they were president, meaning that if they only served as president for one term they had four speeches, whilt the presidents that served two terms had eight speeches each. The only presidents that didn't follow this pattern of having either eight or four terms were presidents that passed before the end of their term, or presidents that had terms during war time.

#### Number of Speeches per Year

```python
# Hint - Use value counts and sort by years

# Counts
speeches = sou["Year"].value_counts()
# Fix Order
speeches = speeches.reindex(sou["Year"].drop_duplicates())

# Plot
speeches.plot(kind="line")
plt.title("Number of State of the Union Speeches per Year")
plt.xlabel("Year")
plt.ylabel("Count")
plt.tight_layout()
plt.show();
```

### Commentary

This plot shows us that most of the time there was only one State of the Union Speech per year, and there were only multiple speeches per year under certain conditions.

#### Word Count Distribution 

```python
# Hint - try seaborn.histplot()

sns.histplot(data = sou['Word Count']).set_title('Distribution of State of the Union Speech Word Counts')
plt.show()
```

### Commentary

We thought it was interesting that this plot didn't follow a bell curve, meaning that most speeches were around a certain word count (around 5000), and that the mean and median would differ greatly.

#### Word Count Distribution over Year

```python
# Hint: try seaborn.rugplot()

sns.scatterplot(data=sou, x="Word Count", y="Year")
sns.rugplot(data=sou, x="Word Count", y="Year")
plt.title('Speech Year Versus Word Count')
plt.show()
```

### Commentary

The evenly distributed ticks on the x-axis shows that there are an equal distribution of speeches per year, while the uneven amount of tickets on the x-axis shows how most speeches had a word count of around 5000 (more concentrated) and very few speeches had a word count above 30000.

#### Word Count Distribution per President

```python
# Hint: use pandas groupby to get mean word count per president then sort by order

# Group by
ave_sou_wc = sou.groupby('President')['Word Count'].mean()
ave_sou_wc = ave_sou_wc.reindex(sou["President"].drop_duplicates())

# Plot
plt.figure(figsize=(12,6))
ave_sou_wc.plot(kind="bar")
plt.title("Average State of the Union Word Count per President")
plt.xlabel("President")
plt.ylabel("Average Word Count")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show();
```

### Commentary

It was interesting to see that even though George Washington made 12 State of the Union speeches during his 8 years as president, his speeches were much shorter than the later speeches. As the Presidents who had a higher average word count aren't evenly distributed in the plot, it's possible that having longer or wordier speeches was a trend.
